## 1. Introducción Teórica

<details>
<summary>1.1 Ensemble Learning</summary>

En Machine Learning, un solo modelo puede ser limitado:  
- Algunos son demasiado simples y no capturan patrones complejos.  
- Otros se ajustan demasiado a los datos de entrenamiento y no generalizan bien.  

El **Ensemble Learning** combina múltiples modelos para construir un predictor más robusto y preciso.  

> “La combinación de varios modelos débiles puede producir un modelo fuerte.”
</details>

---

<details>
<summary>1.2 ¿Por qué utilizar Ensembles?</summary>

Muchos algoritmos, como los árboles de decisión, son muy sensibles a pequeñas variaciones en los datos:  
- Cambios en pocas muestras pueden alterar mucho el modelo.  
- Tienden a sobreajustar.  
- Presentan alta varianza.  

Los métodos ensemble reducen estos problemas al combinar múltiples modelos.
</details>

---

<details>
<summary>1.3 Bagging</summary>

**Bagging (Bootstrap Aggregating)** entrena varios modelos de manera independiente usando diferentes subconjuntos aleatorios del dataset.  

Cada subconjunto se genera con **Bootstrap Sampling**:  
- Selección aleatoria de muestras.  
- Con reemplazo.  

Las predicciones se combinan mediante:  
- Votación (clasificación).  
- Promedio (regresión).  
</details>

---

<details>
<summary>Intuición de Bagging</summary>

Si un modelo individual tiene alta varianza, combinar múltiples modelos independientes puede estabilizar las predicciones.  
En otras palabras, Bagging ayuda a reducir la varianza y mejorar la robustez del modelo.
</details>

---

<details>
<summary>Varianza en Bagging</summary>

La reducción de varianza en Bagging puede expresarse como:



$[
Var(\bar{X}) = \frac{Var(X)}{n}
]$



- $Var(X)$ representa la varianza de un modelo individual.  
- $n$ es la cantidad de modelos utilizados en el ensemble.  

A medida que aumenta $n$:  
- disminuye la varianza,  
- aumenta la estabilidad,  
- mejora la capacidad de generalización.  
</details>

---

<details>
<summary>Características principales de Bagging</summary>

- Entrenamiento paralelo.  
- Modelos independientes.  
- Reduce principalmente la varianza.  
- Funciona bien con modelos inestables.  
- Muy utilizado con árboles de decisión.

</details>



---

<details>
<summary>1.4 Boosting</summary>

El **Boosting** es una estrategia secuencial donde cada nuevo modelo intenta corregir los errores de los anteriores.  

A diferencia de Bagging:  
- los modelos **no son independientes**,  
- cada iteración depende del desempeño previo.  

El objetivo es enfocarse progresivamente en las muestras difíciles de clasificar.  
</details>

---

<details>
<summary>AdaBoost</summary>

AdaBoost asigna pesos a las muestras:  
- Inicialmente todas tienen el mismo peso.  
- Las muestras mal clasificadas aumentan su peso.  
- Las correctamente clasificadas disminuyen su influencia relativa.  

Así, el siguiente clasificador presta mayor atención a los ejemplos difíciles.  
</details>

---

<details>
<summary>Peso del clasificador en AdaBoost</summary>

La importancia de cada clasificador se calcula como:



$[
\alpha_t = \frac{1}{2} \ln \left( \frac{1 - \epsilon_t}{\epsilon_t} \right)
]$



Donde:  
- $\epsilon_t$ es el error del clasificador.  
- $\alpha_t$ representa la importancia del clasificador.  

Si un clasificador tiene bajo error:  
- recibe mayor peso,  
- influye más en la predicción final.  
</details>

---

<details>
<summary>Características principales de Boosting</summary>

- Entrenamiento secuencial.  
- Modelos dependientes entre sí.  
- Se enfoca en corregir errores.  
- Reduce principalmente el sesgo.  
- Puede producir modelos altamente precisos.

<details>
<summary>1.5 Bagging vs Boosting</summary>

| Característica            | Bagging          | Boosting         |
|----------------------------|------------------|------------------|
| Entrenamiento              | Paralelo         | Secuencial       |
| Relación entre modelos     | Independientes   | Dependientes     |
| Objetivo principal         | Reducir varianza | Reducir sesgo    |
| Sensibilidad al ruido      | Menor            | Mayor            |
| Riesgo de sobreajuste      | Bajo             | Puede aumentar   |
| Ejemplo típico             | Random Forest    | AdaBoost         |
</details>

```python
from sklearn.datasets import load_breast_cancer

import pandas as pd
import numpy as np

# ============================================================
# Cargar dataset original
# ============================================================

data = load_breast_cancer()

X = data.data
y = data.target

# Crear DataFrame
df = pd.DataFrame(X, columns=data.feature_names)

# Agregar target
df['target'] = y

print("Dataset original")
print("Shape:", df.shape)
df.head()
```

### Agregar ruido al dataset

```python

np.random.seed(42)

# Nivel de ruido
noise_factor = 0.3

# Generar ruido solo para features
noise = np.random.normal(
    loc=0,
    scale=noise_factor,
    size=X.shape
)

# Aplicar ruido
X_noisy = X + noise

# Actualizar DataFrame
df_noisy = pd.DataFrame(
    X_noisy,
    columns=data.feature_names
)

df_noisy['target'] = y

print("\nRuido gaussiano agregado.")

# Porcentaje de missing values
missing_rate = 0.05

# Cantidad total de valores faltantes
n_missing = int(X.shape[0] * X.shape[1] * missing_rate)

# Índices aleatorios
rows = np.random.randint(0, X.shape[0], n_missing)
cols = np.random.randint(0, X.shape[1], n_missing)

# Insertar NaN
for row, col in zip(rows, cols):
    df_noisy.iat[row, col] = np.nan

print("Valores faltantes agregados.")

# Cantidad de columnas irrelevantes
n_random_features = 15

# Features aleatorios
random_features = np.random.random(
    size=(X.shape[0], n_random_features)
)

# Agregar columnas al DataFrame
for i in range(n_random_features):

    df_noisy[f'random_feature_{i}'] = random_features[:, i]

print("Features irrelevantes agregados.")

# Cantidad de outliers
n_outliers = 30

# Índices aleatorios
outlier_rows = np.random.randint(0, X.shape[0], n_outliers)
outlier_cols = np.random.randint(0, X.shape[1], n_outliers)

# Multiplicar valores para crear outliers
for row, col in zip(outlier_rows, outlier_cols):

    if not pd.isna(df_noisy.iat[row, col]):

        df_noisy.iat[row, col] *= 8

print("Outliers agregados.")

print("\nDataset modificado")
print("Shape final:", df_noisy.shape)

print("\nCantidad de valores faltantes:")
print(df_noisy.isna().sum().sum())

print("\nPrimeras filas del dataset modificado:")

df_noisy.head()
```

### 3.1 Modelo Baseline: Decision Tree

Comenzaremos entrenando un árbol de decisión simple que será utilizado como modelo baseline.

Este modelo servirá como punto de comparación para evaluar posteriormente el impacto de Bagging y Boosting.

#### Configuraciones y Entrenamiento

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, recall_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay

decision_tree = DecisionTreeClassifier(
    criterion="gini",       # Función para medir la calidad de la división (puede ser "gini" o "entropy").
    splitter="best",        # Estrategia para elegir la división en cada nodo ("best" o "random").
    max_depth=30,            # Profundidad máxima del árbol (controla el sobreajuste).
    # min_samples_split=2,    # Número mínimo de muestras necesarias para dividir un nodo.
    # min_samples_leaf=1,     # Número mínimo de muestras que debe tener una hoja.
    # min_weight_fraction_leaf=0.0, # Fracción mínima de peso que debe tener una hoja (útil con datos ponderados).
    # max_features=None,      # Número máximo de características a considerar en cada división.
    random_state=42,        # Semilla para reproducibilidad.
    # max_leaf_nodes=None,    # Número máximo de hojas permitidas en el árbol.
    # min_impurity_decrease=0.0, # Umbral mínimo de disminución de impureza para dividir un nodo.
    # class_weight=None,      # Permite asignar pesos a las clases (útil en datasets desbalanceados).
    # ccp_alpha=0.0           # Parámetro de complejidad para la poda mínima de coste.
)

# Entrenamiento
decision_tree.fit(X_train, y_train)

# Predicciones
y_pred_tree = decision_tree.predict(X_test)
```

#### Evaluación

```python
accuracy_tree = accuracy_score(y_test, y_pred_tree)
recall_score_tree = recall_score(y_test, y_pred_tree)
print("Accuracy Decision Tree:", accuracy_tree)

print(classification_report(y_test, y_pred_tree))

cm_tree = confusion_matrix(y_test, y_pred_tree)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=data.target_names
)

disp.plot(cmap="Blues")

plt.title("Decision Tree - Matriz de Confusión")

plt.show()
```

### 3.2 Bagging

Bagging entrenará múltiples árboles de decisión utilizando subconjuntos aleatorios del dataset.

El objetivo principal es reducir la varianza y aumentar la estabilidad del modelo.

#### Configuraciones y Entrenamiento

```python
from sklearn.ensemble import BaggingClassifier
bagging_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(max_depth=5), # Modelo base: árbol de decisión con profundidad máxima de 5.
    n_estimators=50,            # Número de modelos (árboles) en el ensemble.
    # max_samples=1.0,            # Proporción de muestras usadas para entrenar cada estimador (1.0 = todo el dataset con reemplazo).
    # max_features=1.0,           # Proporción de características usadas en cada estimador (1.0 = todas las features).
    # bootstrap=True,             # Si se hace muestreo con reemplazo (True) o sin reemplazo (False).
    # bootstrap_features=False,   # Si se hace muestreo con reemplazo sobre las características.
    # oob_score=False,            # Si se calcula el desempeño con muestras fuera de bolsa (out-of-bag).
    # warm_start=False,           # Si se reutilizan estimadores previos al aumentar n_estimators.
    # n_jobs=-1,                  # Número de núcleos de CPU a usar (-1 = todos).
    random_state=42,            # Semilla para reproducibilidad.
    # verbose=0                   # Nivel de detalle en la salida (0 = silencioso).
)

# Entrenamiento
bagging_model.fit(X_train, y_train)

# Predicciones
y_pred_bagging = bagging_model.predict(X_test)
```

#### Evaluación

```python
accuracy_bagging = accuracy_score(y_test, y_pred_bagging)
recall_score_bagging = recall_score(y_test, y_pred_bagging)
print("Accuracy Bagging:", accuracy_bagging)


print(classification_report(y_test, y_pred_bagging))

cm_bagging = confusion_matrix(y_test, y_pred_bagging)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_bagging,
    display_labels=data.target_names
)

disp.plot(cmap="Greens")

plt.title("Bagging - Matriz de Confusión")

plt.show()
```

### 3.3 AdaBoost

AdaBoost entrena clasificadores secuencialmente enfocándose en las muestras más difíciles de clasificar.

Cada clasificador intenta corregir los errores cometidos por el anterior.

#### Configuraciones y Entrenamiento

```python
from sklearn.ensemble import AdaBoostClassifier
ada_model = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=5), # Modelo base: árbol de decisión con profundidad máxima de 5.
    n_estimators=50,            # Número de clasificadores secuenciales que se entrenan.
    # learning_rate=1.0,          # Factor de ajuste de la contribución de cada clasificador (menor valor = más robusto, pero más lento).
    # algorithm="SAMME.R",        # Algoritmo de boosting: "SAMME.R" (real boosting, más común) o "SAMME".
    random_state=42             # Semilla para reproducibilidad.
)

# Entrenamiento
ada_model.fit(X_train, y_train)

# Predicciones
y_pred_ada = ada_model.predict(X_test)
```

#### Evaluación

```python
accuracy_ada = accuracy_score(y_test, y_pred_ada)
recall_score_ada = recall_score(y_test, y_pred_ada)
print("Accuracy AdaBoost:", accuracy_ada)


print(classification_report(y_test, y_pred_ada))

cm_ada = confusion_matrix(y_test, y_pred_ada)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_ada,
    display_labels=data.target_names
)

disp.plot(cmap="Oranges")

plt.title("AdaBoost - Matriz de Confusión")

plt.show()
```

### 3.4 Gradient Boosting

Gradient Boosting optimiza progresivamente el error del modelo construyendo árboles secuencialmente.

Cada nuevo árbol intenta corregir el error residual del sistema.

#### Configuraciones y Entrenamiento

```python
from sklearn.ensemble import GradientBoostingClassifier
gb_model = GradientBoostingClassifier(
    # loss="log_loss",          # Función de pérdida: "log_loss" para clasificación binaria/multiclase.
    # learning_rate=0.1,        # Tasa de aprendizaje: controla cuánto contribuye cada árbol (menor = más robusto, pero más lento).
    n_estimators=100,         # Número de árboles secuenciales en el ensemble.
    # subsample=1.0,            # Proporción de muestras usadas para entrenar cada árbol (menor a 1.0 introduce aleatoriedad).
    # criterion="friedman_mse", # Criterio para medir la calidad de la división ("friedman_mse", "squared_error", "mse").
    # min_samples_split=2,      # Número mínimo de muestras necesarias para dividir un nodo.
    # min_samples_leaf=1,       # Número mínimo de muestras en una hoja.
    # min_weight_fraction_leaf=0.0, # Fracción mínima de peso que debe tener una hoja.
    max_depth=3,              # Profundidad máxima de cada árbol (controla el sobreajuste).
    # min_impurity_decrease=0.0,# Umbral mínimo de disminución de impureza para dividir un nodo.
    # max_features=None,        # Número máximo de características consideradas en cada división.
    random_state=42,          # Semilla para reproducibilidad.
    # max_leaf_nodes=None,      # Número máximo de hojas permitidas en cada árbol.
    # warm_start=False,         # Si se reutilizan estimadores previos al aumentar n_estimators.
    # validation_fraction=0.1,  # Proporción de datos usados para validación temprana (early stopping).
    # n_iter_no_change=None,    # Número de iteraciones sin mejora antes de detener el entrenamiento.
    # tol=1e-4,                 # Tolerancia mínima de mejora para early stopping.
    # verbose=0                 
)
# Entrenamiento
gb_model.fit(X_train, y_train)

# Predicciones
y_pred_gb = gb_model.predict(X_test)
```

#### Evaluación

```python
accuracy_gb = accuracy_score(y_test, y_pred_gb)
recall_score_gb = recall_score(y_test, y_pred_gb)
print("Accuracy Gradient Boosting:", accuracy_gb)

print(classification_report(y_test, y_pred_gb))

cm_gb = confusion_matrix(y_test, y_pred_gb)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_gb,
    display_labels=data.target_names
)

disp.plot(cmap="Purples")

plt.title("Gradient Boosting - Matriz de Confusión")

plt.show()
```
